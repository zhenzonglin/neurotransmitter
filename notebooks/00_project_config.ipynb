# Project Configuration

Edit this notebook first. It writes `config/dat_config.yaml`, and all work scripts receive the settings through `--config config/dat_config.yaml`.

```python
from pathlib import Path
import yaml

# 只在这里修改工作路径
project_dir = Path('/home/zhenzong2/analysis/neurotransmitter')
config_path = project_dir / 'config' / 'dat_config.yaml'

# 只在这里修改原始数据入口
lesion_glob = 'lesion/*.nii*'
phenotype_file = 'phenotype.xlsx'
phenotype_sheet = 'Sheet1'
lesion_subject_regex = r'sub-(TMS\d{3})ses\d+'
phenotype_id_column = 'ID'

# 只在这里修改结局和临床变量
raw_outcome_column = 'mRS_3m'
outcome = 'mrs_3m'
covariate_columns = {
    'age': '年龄',
    'sex': '性别',
    'nihss': 'NIHSS',
}
categorical_maps = {
    'sex': {'男': 1, '女': 0, '1': 1, '0': 0},
}

# NiiStat和整合模型可使用不同协变量
niistat_covariates = ['age', 'sex', 'nihss']
model_covariates = ['age', 'sex', 'nihss', 'lesion_volume_ml']

# 二分类敏感性分析定义
binary_threshold = 2
positive_if_less_equal = True

# 输出表名不含样本量，便于后续扩展
node_damage_table = 'dat_node_damage.csv'
lqt_edge_table = 'lqt_edge_disconnection.csv'
dat_edge_table = 'dat_edge_lqt.csv'

# DAT impact score分析设置
impact_n_splits = 10
impact_random_state = 42
impact_weight_stat = 't'
impact_q_threshold = 0.05
impact_node_top_k = 20
impact_edge_top_k = 200
impact_use_q_if_available = True
impact_dataset_name = 'main'
prediction_bootstrap = 1000
```

## Stable Resource Paths

These paths usually stay fixed after the project is initialized. Change them here if the atlas, LQT, or NiiStat resources move.

```python
# 图谱和外部项目路径
atlas_raw_dir = 'atlases/raw'
atlas_processed_dir = 'atlases/processed'
atlas_lqt_dir = 'atlases/lqt'

atlas4s156_nii = 'atlases/raw/tpl-MNI152NLin6Asym_atlas-4S156Parcels_res-01_dseg.nii.gz'
atlas4s156_tsv = 'atlases/raw/atlas-4S156Parcels_dseg.tsv'
atlas4s156_json = 'atlases/raw/tpl-MNI152NLin6Asym_atlas-4S156Parcels_dseg.json'
hansen_dat_nii = 'atlases/raw/DAT_fpcit_hc174_dukart_spect.nii'
dat_wm_nii = 'atlases/raw/functionnectome_anat_DAT.nii.gz'

lqt_repo_dir = 'external/lqt'
lqt_data_dir = 'external/lqt_data'
dsi_path = 'scripts/dsi_studio_docker.sh'
docker_mount_root = str(project_dir.parents[1])

niistat_repo_dir = 'external/niistat'
matlab_bin = 'matlab'

# 派生结果目录
qc_dir = 'derivatives/qc'
node_dir = 'derivatives/node_clsm'
wm_voxel_dir = 'derivatives/wm_voxel_clsm'
edge_dir = 'derivatives/edge_clsm'
model_dir = 'derivatives/models'
impact_dir = 'derivatives/impact'
```

## Write Config

Run this cell after editing variables. It grants the settings to all scripts by rewriting `config/dat_config.yaml`.

```python
config = {
    'project_dir': str(project_dir),
    'inputs': {
        'lesion_glob': lesion_glob,
        'phenotype_file': phenotype_file,
        'phenotype_sheet': phenotype_sheet,
        'lesion_subject_regex': lesion_subject_regex,
        'phenotype_id_column': phenotype_id_column,
        'outcome_column': raw_outcome_column,
        'covariates': covariate_columns,
        'categorical_maps': categorical_maps,
    },
    'analysis': {
        'outcome': outcome,
        'niistat_covariates': niistat_covariates,
        'model_covariates': model_covariates,
        'binary_outcome': {
            'threshold': binary_threshold,
            'positive_if_less_equal': positive_if_less_equal,
        },
        'tables': {
            'node_damage': node_damage_table,
            'lqt_edge_disconnection': lqt_edge_table,
            'dat_edge_lqt': dat_edge_table,
        },
    },
    'impact': {
        'n_splits': impact_n_splits,
        'random_state': impact_random_state,
        'dataset_name': impact_dataset_name,
        'weight_stat': impact_weight_stat,
        'q_threshold': impact_q_threshold,
        'node_top_k': impact_node_top_k,
        'edge_top_k': impact_edge_top_k,
        'use_q_if_available': impact_use_q_if_available,
        'prediction_bootstrap': prediction_bootstrap,
    },
    'atlases': {
        'raw_dir': atlas_raw_dir,
        'processed_dir': atlas_processed_dir,
        'lqt_dir': atlas_lqt_dir,
        'atlas4s156': {
            'nii': atlas4s156_nii,
            'tsv': atlas4s156_tsv,
            'json': atlas4s156_json,
        },
        'hansen_dat': {'nii': hansen_dat_nii},
        'dat_wm': {'nii': dat_wm_nii},
        'outputs': {
            'atlas4s156_2mm': 'atlases/processed/atlas4s156_2mm.nii.gz',
            'atlas4s156_1mm_lqt': 'atlases/lqt/atlas4s156_1mm.nii.gz',
            'dat_gray_2mm': 'atlases/processed/dat_gray_2mm.nii.gz',
            'dat_wm_2mm': 'atlases/processed/dat_wm_2mm.nii.gz',
            'dat_wm_mask_2mm': 'atlases/processed/dat_wm_mask_2mm.nii.gz',
            'atlas4s156_lqt_nodes': 'atlases/processed/atlas4s156_lqt_nodes.csv',
        },
    },
    'lqt': {
        'repo_dir': lqt_repo_dir,
        'data_dir': lqt_data_dir,
        'dsi_path': dsi_path,
        'docker_mount_root': docker_mount_root,
        'con_type': 'end',
        'file_suffix': 'atlas4s156',
        'max_subjects': None,
    },
    'niistat': {
        'matlab_bin': matlab_bin,
        'repo_dir': niistat_repo_dir,
        'num_permute': -1,
        'p_thresh': 0.05,
        'min_overlap': 3,
    },
    'outputs': {
        'qc_dir': qc_dir,
        'node_dir': node_dir,
        'wm_voxel_dir': wm_voxel_dir,
        'edge_dir': edge_dir,
        'model_dir': model_dir,
        'impact_dir': impact_dir,
    },
}

# 写出脚本读取的唯一配置文件
config_path.parent.mkdir(parents=True, exist_ok=True)
with config_path.open('w', encoding='utf-8') as f:
    yaml.safe_dump(config, f, sort_keys=False, allow_unicode=True)

print(f'wrote {config_path}')
```
